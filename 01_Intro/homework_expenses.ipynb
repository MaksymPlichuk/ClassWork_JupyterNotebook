```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]
total_expenses = sum(expenses)
avg_expense = total_expenses / len(expenses)
max_expense = max(expenses)
min_expense = min(expenses)

days_above_1000 = 0
for exp in expenses:
    if exp > 1000:
        days_above_1000+=1

markdown_table = f"""
| Метрика витрат | Значення |
| :--- | :--- |
| **Загальна сума** | {total_expenses} грн |
| **Середня витрата на день** | {avg_expense:.2f} грн |
| **Максимальна витрата** | {max_expense} грн |
| **Мінімальна витрата** | {min_expense} грн |
| **Днів із витратами > 1000 грн** | {days_above_1000} |
"""

from IPython.display import display, Markdown
display(Markdown(markdown_table))
```
