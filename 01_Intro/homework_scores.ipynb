```python
scores = {
    "Math": [80, 92, 75, 60, 88],
    "Physics": [70, 65, 90, 85, 78],
    "English": [95, 100, 85, 90, 87]
}
results = {}
for subject, marks in scores.items():
    results[subject] = {
        "avg": sum(marks) / len(marks),
        "min": min(marks),
        "max": max(marks)
    }
best_subject = max(results, key=lambda subj: results[subj]["avg"])

table_rows = "\n".join([
    f"| {subj} | {data['avg']:.2f} | {data['min']} | {data['max']} |"
    for subj, data in results.items()
])

markdown_table = f"""
| Предмет | Середній бал | Мінімальний бал | Максимальний бал |
| :--- | :---: | :---: | :---: |
{table_rows}

**Предмет з найвищим середнім результатом:** `{best_subject}` ({results[best_subject]['avg']:.2f})
"""

from IPython.display import display, Markdown
display(Markdown(markdown_table))
```
