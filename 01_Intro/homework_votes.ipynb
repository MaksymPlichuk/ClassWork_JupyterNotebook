```python
from collections import Counter
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
vote_counts = Counter(votes)

leader, leader_votes = vote_counts.most_common(1)[0]
top_3 = vote_counts.most_common(3)

table_rows = "\n".join([f"| {fruit} | {count} |" for fruit, count in vote_counts.items()])
top3_rows = "\n".join([f"| {i+1} | {fruit} | {count} |" for i, (fruit, count) in enumerate(top_3)])

markdown_output = f"""
### Загальні результати голосування
| Фрукт | Кількість голосів |
| :--- | :--- |
{table_rows}

**Переможець:** `{leader}` ({leader_votes} голосів)

---

### Топ-3 фрукти
| Місце | Фрукт | Кількість голосів |
| :---: | :--- | :--- |
{top3_rows}
"""

from IPython.display import display, Markdown
display(Markdown(markdown_output))
```
